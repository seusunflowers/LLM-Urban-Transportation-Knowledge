Inspired by the vast knowledge parameterized by the LLMs, we assume they can comprehensively understand the spatial and temporal profiles of urban transportation systems reported in corpora. A promising method for validation is to decode the spatial and temporal features from their hidden states. Deterministic attributes such as geographic locations, POI patterns, and time-of-day indices are considered as they contribute to the fundamental spatiotemporal profiles of transportation systems and underpin complex traffic dynamics. Linear regression models (least squares) are applied to extract the semantic information embedded in **Qwen-3-14B**’s PCA-condensed hidden states. Significant accuracy and statistical correlation between the decoded and true features would provide evidence for the assumption.

The core logic is as follows:
* **Embedding Extraction** Hidden states (typically from a Qwen3 model) are loaded from `.npz` files. We apply PCA to condense the high-dimensional LLM output (e.g., 5120 dimensions) into manageable components (e.g., 20–80 dimensions) while retaining maximum variance.
* **Feature Alignment** Real-world features are prepared as targets:
   - **Spatial**: Coordinates (X, Y) and POI distributions (via OpenStreetMap).
   - **Structural**: Graph topology represented by Node2Vec embeddings.
   - **Temporal**: Time-of-day (cosine encoding) and day-of-week.
   - **Observational**: Historical average traffic speed, ridership, or occupancy.
* **Linear Decoding** We use a linear probe to see if a simple transformation can reconstruct real-world features from LLM states. High correlation ($R_s, R_p$) indicates that the LLM has successfully "encoded" that specific urban concept in its latent space.
* **Evaluation Metrics** The performance is measured using:
   - **MAE**: Average error in the reconstructed normalized feature.
   - **Spearman ($R_s$)**: Rank correlation between decoded and real values.
   - **Pearson ($R_p$)**: Linear correlation between decoded and real values.
* **Visualizations** The script generates several plots, including:
   - **Geographical KDE Maps**: Density of decoded station locations vs. real locations.
   - **Topology Links**: Comparison between real graph adjacency and decoded correlations.
   - **Trend Comparison**: Decoded weekly/daily trends compared against real historical observations.

```python
def get_spatiotemporal_embeddings(hidden_state_dir:str, node_pca_dim:int, time_pca_dim:int, day_pca_dim:int, time_repeats:int=7, node_type:str='station',
                                  interp_time_steps:int=24):
    ''' Loads raw hidden states from a directory and applies Principal Component Analysis (PCA) to reduce the dimensionality of node, time, and day embeddings.
        It also handles linear interpolation for time steps to match specific dataset frequencies.
    '''
    raw_embeddings = np.load(hidden_state_dir)
    raw_node_embeddings, raw_time_embeddings, raw_day_embeddings = raw_embeddings[node_type], raw_embeddings['time'], raw_embeddings['day']
    # print(raw_node_embeddings.shape, raw_time_embeddings.shape, raw_day_embeddings.shape)
    node_embeddings, time_embeddings, day_embeddings = raw_node_embeddings[:,-1,:], raw_time_embeddings[:,-1,:], raw_day_embeddings[:,-1,:]
    
    pca = PCA(n_components=node_pca_dim, random_state=42)
    node_embeddings = pca.fit_transform(node_embeddings)
    print(pca.explained_variance_ratio_.sum())
    node_embeddings = torch.from_numpy(node_embeddings).type(torch.FloatTensor).to(device)
    node_embeddings = (node_embeddings - node_embeddings.mean()) / node_embeddings.std()
    
    pca = PCA(n_components=time_pca_dim, random_state=42)
    time_embeddings = pca.fit_transform(time_embeddings)
    print(pca.explained_variance_ratio_.sum())
    raw_time_steps = len(time_embeddings)
    if interp_time_steps != raw_time_steps:
        assert interp_time_steps > 0
        time_embeddings = np.concatenate((time_embeddings, time_embeddings[:1,:]))
        f = scipy.interpolate.interp1d(np.arange(raw_time_steps + 1) * interp_time_steps / raw_time_steps, time_embeddings, axis=0, kind='linear',
                                       fill_value="extrapolate")
        time_embeddings = f(np.arange(interp_time_steps))
    time_embeddings = torch.from_numpy(time_embeddings).type(torch.FloatTensor).to(device)
    
    pca = PCA(n_components=day_pca_dim, random_state=42)
    day_embeddings = pca.fit_transform(day_embeddings)
    print(pca.explained_variance_ratio_.sum())
    day_embeddings = torch.from_numpy(day_embeddings).type(torch.FloatTensor).to(device)
    
    time_embeddings = time_embeddings.repeat(time_repeats, 1)
    day_embeddings = day_embeddings.unsqueeze(dim=1).repeat(1,DAILY_TIME_STEPS,1).view(time_repeats*DAILY_TIME_STEPS, -1)
    time_embeddings = torch.cat([time_embeddings, day_embeddings], dim=1)
    time_embeddings = (time_embeddings - time_embeddings.mean()) / time_embeddings.std()
    
    return node_embeddings, time_embeddings
```

```python
def calculate_linear_reconstruction_scores(input_e, target_e): 
    ''' The core "decoding" engine. It fits a Linear Regression model to map input embeddings to target features, 
        returning accuracy metrics like Mean Absolute Error (MAE), Spearman Correlation ($R_s$), and Pearson Correlation ($R_p$).
    '''
    from sklearn.linear_model import LinearRegression
    
    e_num, target_e_dim = target_e.shape
    input_mean, input_std = input_e.mean(axis=0), input_e.std(axis=0)
    target_mean, target_std = target_e.mean(axis=0), target_e.std(axis=0)
    input_e = (input_e - input_mean) / input_std
    target_e = (target_e - target_mean) / target_std
    model = LinearRegression()
    model.fit(input_e, target_e)
    Y_hat = model.predict(input_e)
    mae = (np.abs(Y_hat-target_e)).mean()
    spearman_value = sum([scipy.stats.spearmanr(Y_hat[:,i], target_e[:,i]).statistic for i in range(target_e_dim)]) / target_e_dim
    pearson_value = sum([scipy.stats.pearsonr(Y_hat[:,i], target_e[:,i]).statistic for i in range(target_e_dim)]) / target_e_dim
    return {'MAE':mae, 'rs':spearman_value, 'rp':pearson_value}, Y_hat * target_std + target_mean

def semantic_feature_reconstruction(input_type_list:list, input_e_list:list, target_type_list:list, target_e_list:list): 
    ''' A wrapper function that iterates through lists of input types (e.g., PCA-reduced LLM states) and target types (e.g., Location, POI, Topology) 
        to perform batch decoding and store results.
    '''
    decoding_results = {}
    for input_type, input_e in zip(input_type_list, input_e_list):
        decoding_results[input_type] = {'Score':{}, 'Y_hat':{}, 'Y':{}}
        for target_type, target_e in zip(target_type_list, target_e_list):
            scores, y_hat = calculate_linear_reconstruction_scores(input_e, target_e)
            print(input_type, target_type, scores)
            decoding_results[input_type]['Score'][target_type] = scores
            decoding_results[input_type]['Y_hat'][target_type] = y_hat
            decoding_results[input_type]['Y'][target_type] = target_e.copy()
    return decoding_results
```

```python
raw_data = np.load(f'{RAW_DATASET_DIR}/station-based-st-graph.npz', allow_pickle=True)
# X_bike, X_weather = raw_data['bike_data'], raw_data['weather_data']
X_trip = raw_data['ridership_data']
time_cos = np.cos(X_trip[:,1:2,:] * 2 * np.pi / DAILY_TIME_STEPS)
is_saturday = (X_trip[:,2:,:]==5).astype(int)
is_sunday = (X_trip[:,2:,:]==6).astype(int)
X = np.concatenate((X_trip[:,0:1,:], time_cos, is_saturday, is_sunday), axis=1)
data_loader = MyStaticSTGDatasetLoader(X=X.astype(np.float32), A=raw_data['A'].astype(np.float32))
_, _, TOTAL_TIME_STEPS = data_loader.X.shape
y_mean, y_std = data_loader.X[:,0,:].mean(), data_loader.X[:,0,:].std()
data_loader.X[:,0,:] = (data_loader.X[:,0,:] - y_mean) / y_std # Normalization

node_embedding_list, time_embedding_list = [], []
for hidden_state_name in ['description-qwen3-14b/hidden_states_qwen3_14b.npz']:
    node_embeddings, time_embeddings = get_spatiotemporal_embeddings(hidden_state_dir=f'{RAW_DATASET_DIR}/{hidden_state_name}',
                                                                     node_pca_dim=45, time_pca_dim=20, day_pca_dim=5, time_repeats=7, node_type='station')
    node_embedding_list.append(node_embeddings)
    time_embedding_list.append(time_embeddings)

NODE_NUM, INIT_NODE_E_DIM = node_embeddings.shape
_, INIT_TIME_E_DIM = time_embeddings.shape
```

## Linear-Feature-Reconstruction

```python
pca_time_e = time_embedding_list[0].detach().cpu().numpy()

time_stamps = np.concatenate((np.array([0] * (5 * DAILY_TIME_STEPS) + [1] * (2 * DAILY_TIME_STEPS))[:,True],
                              np.cos(np.repeat(np.arange(DAILY_TIME_STEPS)[True,:], 7, axis=0).flatten() / DAILY_TIME_STEPS * 2 * np.pi)[:,True]), axis=1)
weekly_x = DataFrame(data=data_loader.X[:,0,:].permute(1,0).detach().numpy()).groupby(by=[X_trip[0,2,:].astype(int), X_trip[0,1,:].astype(int)]).mean().values * y_std.item() + y_mean.item()
```

```python
## Temporal Feature Reconstruction
bart_time_results = semantic_feature_reconstruction(input_type_list=['PCA'], input_e_list=[pca_time_e], target_type_list=['Time', 'Observation'], target_e_list=[time_stamps, weekly_x])
```

```python
from sklearn.decomposition import PCA

## electric and service prices yuan/kWh
price = pd.read_csv(f'{RAW_DATASET_DIR}/s_price.csv').iloc[:,1:1+NODE_NUM] + \
        pd.read_csv(f'{RAW_DATASET_DIR}/e_price.csv').iloc[:,1:1+NODE_NUM]
price = price / price.max()
## weather information
weather = pd.read_csv(f'{RAW_DATASET_DIR}/20220901-20230228_weather_central.csv', usecols=[1,2,3,4,5,6])
weather = (weather - weather.mean()) / weather.std()
## charging station occupancy
raw_data = pd.read_csv(f'{RAW_DATASET_DIR}/occupancy.csv')
raw_data['time'] = pd.to_datetime(raw_data['time'])
raw_data['hour'] = raw_data['time'].dt.hour
raw_data['dayoff'] = (~raw_data['time'].map(chinese_calendar.is_workday)).astype(int)
raw_data['predayoff'] = (~raw_data['time'].map(lambda dt: chinese_calendar.is_workday(dt + datetime.timedelta(days=1)))).astype(int)
X = np.concatenate((raw_data.iloc[:,1:1+NODE_NUM].values[:,True,:].T,
                    np.repeat(np.cos(raw_data['hour'] * 2 * np.pi / DAILY_TIME_STEPS).values[True,:], NODE_NUM, axis=0)[:,True,:],
                    np.repeat(raw_data.iloc[:,2+NODE_NUM:].T.values[True,:,:], NODE_NUM, axis=0),
                    price.values[:,True,:].T,
                    np.repeat(weather.T.values[True,:,:], NODE_NUM, axis=0)), axis=1)
# generate adjacency matrix
pre_A = pd.read_csv(f'{RAW_DATASET_DIR}/adj.csv').values
zone_dist = pd.read_csv(f'{RAW_DATASET_DIR}/distance.csv').values
zone_dist[(zone_dist > np.percentile(zone_dist, 5)) & (pre_A == 0)] = np.nan
A = np.exp(-(zone_dist / np.nanstd(zone_dist))**2 * 0.3)
A[np.isnan(A)] = 0
A[(A < 0.05) & (A > 0)] = 0.05
data_loader = MyStaticSTGDatasetLoader(X=X.astype(np.float32), A=A.astype(np.float32))
_, _, TOTAL_TIME_STEPS = data_loader.X.shape
y_mean, y_std = data_loader.X[:,0,:].mean(), data_loader.X[:,0,:].std()
data_loader.X[:,0,:] = (data_loader.X[:,0,:] - y_mean) / y_std # Normalization
print(y_mean, y_std)

node_embedding_list, time_embedding_list = [], []
for hidden_state_name in ['description-qwen3-14b/hidden_states_qwen3_14b.npz']:
    node_embeddings, time_embeddings = get_spatiotemporal_embeddings(hidden_state_dir=f'{RAW_DATASET_DIR}/{hidden_state_name}',
                                                                     node_pca_dim=80, time_pca_dim=20, day_pca_dim=6, time_repeats=8, node_type='zone')
    node_embedding_list.append(node_embeddings)
    time_embedding_list.append(time_embeddings)

_, INIT_NODE_E_DIM = node_embeddings.shape
_, INIT_TIME_E_DIM = time_embeddings.shape
```

```python
def truncate_colormap(cmap, minval=0.0, maxval=1.0, n=-1):
    if n == -1:
        n = cmap.N
    new_cmap = mpl.colors.LinearSegmentedColormap.from_list(
         'trunc({name},{a:.2f},{b:.2f})'.format(name=cmap.name, a=minval, b=maxval),
         cmap(np.linspace(minval, maxval, n)))
    return new_cmap
```

```python
def cal_knn_density(train_data, xy_grid, K, P):
    from sklearn.neighbors import NearestNeighbors
    
    knn = NearestNeighbors(n_neighbors=K)
    knn.fit(train_data)
    
    distances, _ = knn.kneighbors(xy_grid)
    counts = (distances < np.percentile(distances, P)).sum(axis=1) / K
    distances[distances >= np.percentile(distances, P)] = np.nan
    mean_distances = np.nanmean(distances, axis=1)
    mean_distances[np.isnan(mean_distances)] = np.inf
    densities = counts / mean_distances
    return densities
```

```python
## NYC ridershp data
nyc_raw_data = np.load('subway-mta/station-based-st-graph.npz', allow_pickle=True)
nyc_station_info = pd.read_csv('subway-mta/MTA_Subway_Hourly_Ridership__2020-2024_20250317.zip', usecols=[3, 4, 9, 10])
nyc_station_info.drop_duplicates(subset=['station_complex'], inplace=True)
nyc_station_info = gpd.GeoDataFrame(
    nyc_station_info, geometry=gpd.points_from_xy(nyc_station_info.longitude, nyc_station_info.latitude), crs="EPSG:4326"
).to_crs(epsg=3857)
nyc_station_info.set_index('station_complex', inplace=True)
nyc_station_info = nyc_station_info.loc[nyc_raw_data['station_ids']]
## Shenzhen electric vehicle charging data
sz_zones_shp = gpd.read_file('UrbanEV/SZ_districts/SZ_districts.shp')
sz_zones_shp['TAZID'] = sz_zones_shp['TAZID'].astype(int)
sz_zones_info = pd.read_csv('UrbanEV/zone-information.csv')
sz_zones_shp = sz_zones_shp[sz_zones_shp['TAZID'].isin(sz_zones_info['TAZID'])]
sz_subdistricts_shp = gpd.read_file('UrbanEV/SZ_districts/subdistrict.shp', encoding='utf-8')
sz_subdistricts_shp.rename(columns={'NAME': 'SUBDISCTRIC'}, inplace=True)
sz_subdistricts_shp.to_crs(sz_zones_shp.crs, inplace=True)
sz_sub_zones = gpd.overlay(sz_zones_shp, sz_subdistricts_shp[['SUBDISCTRIC', 'DISTRICT', 'geometry']], how='intersection') 
sz_sub_zones['area_ov'] = sz_sub_zones.geometry.area                                   # For TAZs spanning multiple subdistricts, calculate their area located on each subdistrict
sz_sub_zones.sort_values(by=['OBJECTID', 'area_ov'], inplace=True)                     # For TAZs spanning multiple subdistricts, sort them by the area within each subdistrict in ascending order
sz_sub_zones = sz_sub_zones.drop_duplicates(subset=['OBJECTID', 'TAZID'], keep='last') # For TAZs spanning multiple subdistricts, the affiliated subdistricts\ is determined by the portion with the largest area
sz_zones_shp = sz_zones_shp.merge(right=sz_sub_zones[['OBJECTID', 'TAZID', 'SUBDISCTRIC', 'DISTRICT']], on=['OBJECTID', 'TAZID'], how='left')
sz_zones_shp.sort_values(by='TAZID', inplace=True)
## METR-LA
la_station_info = pd.read_csv('metr-la/graph_sensor_metadata.csv', index_col=0)
la_station_info = gpd.GeoDataFrame(
    la_station_info, geometry=gpd.points_from_xy(la_station_info.longitude, la_station_info.latitude), crs="EPSG:4326"
).to_crs(epsg=3857)

## Visualization
K, P = 5, 25
dataset_names = ['SUBWAY-MTA', 'UrbanEV', 'METR-LA']
label_size = 12
with plt.style.context(['science', 'no-latex']):
    fig, axes = plt.subplots(2, 3, figsize=(12,5), layout='compressed')

    for row in axes:
        for ax in row:
            ax.set_xticks([])
            ax.set_yticks([])

    ## Geographical location
    for i, (dataset_name, node_results) in enumerate(zip(dataset_names, [nyc_node_results, urbanev_node_results, metrla_node_results])):
        cood_pred = node_results['PCA']['Y_hat']['Location']
        cood_true = node_results['PCA']['Y']['Location']
        cood_pred = (cood_pred - cood_true.mean(axis=0)) / cood_true.std(axis=0)
        cood_true = (cood_true - cood_true.mean(axis=0)) / cood_true.std(axis=0)
        x_grid = np.linspace(min(cood_pred[:,0].min(), cood_true[:,0].min()), max(cood_pred[:,0].max(), cood_true[:,0].max()), 100)
        y_grid = np.linspace(min(cood_pred[:,1].min(), cood_true[:,1].min()), max(cood_pred[:,1].max(), cood_true[:,1].max()), 100)
        xy_grid = np.array(np.meshgrid(x_grid, y_grid)).T.reshape(-1, 2)
        densities = cal_knn_density(train_data=cood_pred, xy_grid=xy_grid, K=K, P=P)
        densities = np.log(densities + 1)

        cs = axes[0][i].tricontourf(xy_grid[:,0], xy_grid[:,1], densities, levels=10, cmap=plt.cm.Blues)
        axes[0][i].scatter(cood_pred[:,0], cood_pred[:,1], s=2, label='Linearly Decoded Location')
        axes[0][i].scatter(cood_true[:,0], cood_true[:,1], s=4, marker='s', c='k', label='Real Location')
        axes[0][i].set_title(f"({chr(97+i)}). {dataset_name}", y=-0.12, fontsize=label_size)
        cbar = fig.colorbar(cs, ax=axes[0][i], aspect=20, pad=0.02)
        cbar.ax.tick_params(labelsize=label_size-1)

    ## NYC metro stations in Manhattan and the Bronx
    nyc_borough_bound = gpd.read_file(geodatasets.get_path("nybb")).to_crs(nyc_station_info.crs)
    sub_nyc_borough_bound = nyc_borough_bound[nyc_borough_bound['BoroName'].isin(['Manhattan', 'Bronx'])]
    cood_pred = nyc_node_results['PCA']['Y_hat']['Location']
    sub_nyc_borough_bound.plot(ax=axes[1][0], edgecolor="k", color=['#87ceeb', '#f08080'], alpha=0.2, lw=0.5)
    axes[1][0].scatter(cood_pred[nyc_station_info['borough'] == 'Manhattan',0], cood_pred[nyc_station_info['borough'] == 'Manhattan',1], s=2, label='Station in Manhattan')
    axes[1][0].scatter(cood_pred[nyc_station_info['borough'] == 'Bronx',0], cood_pred[nyc_station_info['borough'] == 'Bronx',1], s=2, c='#8b0000', label='Station in Bronx')
    axes[1][0].legend(loc='upper left', fontsize=label_size-1, markerscale=2, handletextpad=0, fancybox=False, frameon=True)
    axes[1][0].set_xlim(-8.259e6, -8.201e6)
    cx.add_basemap(axes[1][0], crs=sub_nyc_borough_bound.crs, zoom=12, source=cx.providers.CartoDB.VoyagerNoLabels)
    axes[1][0].set_title('(d). Decoded Stations in Manhattan and the Bronx', y=-0.12, fontsize=label_size)

    ## Shenzhen TAZs in Luohu and Futian
    sz_subdistricts_shp = sz_subdistricts_shp[sz_subdistricts_shp['DISTRICT'].isin(['罗湖区', '福田区'])]
    sz_districts_shp = sz_subdistricts_shp[['DISTRICT', 'geometry']].dissolve(by='DISTRICT')
    cood_true = sz_zones_shp['geometry'].centroid
    cood_pred = urbanev_node_results['PCA']['Y_hat']['Location']
    sz_districts_shp.plot(ax=axes[1][1], edgecolor="k", color=['#f08080', '#87ceeb'], alpha=0.2, lw=0.5)
    axes[1][1].scatter(cood_pred[sz_zones_shp['DISTRICT']=='罗湖区',0], cood_pred[sz_zones_shp['DISTRICT']=='罗湖区',1], s=4, label='TAZ in Luohu')
    axes[1][1].scatter(cood_pred[sz_zones_shp['DISTRICT']=='福田区',0], cood_pred[sz_zones_shp['DISTRICT']=='福田区',1], s=4, c='#8b0000', label='TAZ in Futian')
    axes[1][1].legend(loc='upper left', fontsize=label_size-1, markerscale=2, handletextpad=0, fancybox=False, frameon=True)
    axes[1][1].set_xlim(1.268e7, 1.2715e7)
    axes[1][1].set_ylim(2.5665e6, 2.589e6)
    cx.add_basemap(axes[1][1], crs=sz_districts_shp.crs, zoom=13, source=cx.providers.CartoDB.VoyagerNoLabels)
    axes[1][1].set_title('(e). Decoded Centroids of TAZs in Luohu and Futian', y=-0.12, fontsize=label_size)

    ## LA highway sensors on US-101 and SR-134
    cood_pred = metrla_node_results['PCA']['Y_hat']['Location']
    la_station_info[la_station_info['FwyName'].isin(['US-101', 'SR-134'])].plot(ax=axes[1][2], markersize=4, marker='s', color='k')
    axes[1][2].scatter(cood_pred[la_station_info['FwyName'] == 'US-101',0], cood_pred[la_station_info['FwyName'] == 'US-101',1], s=2, label='Detectors on US-101')
    axes[1][2].scatter(cood_pred[la_station_info['FwyName'] == 'SR-134',0], cood_pred[la_station_info['FwyName'] == 'SR-134',1], s=2, c='#8b0000', label='Detectors on SR-134')
    axes[1][2].legend(loc='upper left', fontsize=label_size-1, markerscale=2, handletextpad=0, fancybox=False, frameon=True)
    axes[1][2].set_ylim(4.0295e6, 4.0605e6)
    cx.add_basemap(axes[1][2], crs=la_station_info.crs, zoom=12, source=cx.providers.CartoDB.VoyagerNoLabels)
    axes[1][2].set_title('(f). Decoded Detectors on US-101 and SR-134', y=-0.12, fontsize=label_size)

    cbar.ax.set_ylabel('Logarithmic Density', fontsize=label_size)
    points, labels = axes[0][0].get_legend_handles_labels()
    fig.legend(points, labels, loc='upper center', ncol=2, fontsize=label_size, markerscale=2, handletextpad=0, fancybox=False, frameon=False, bbox_to_anchor=(0.5, 1.05))
    
    plt.show()
```

```python
dataset_names = ['SUBWAY-MTA', 'UrbanEV', 'METR-LA']

label_size = 14
with plt.style.context(['science', 'ieee', 'grid', 'no-latex']):
    fig, axes = plt.subplots(3, 4, figsize=(15,9), layout='compressed')

    for i, (dataset_name, node_results) in enumerate(zip(dataset_names, [nyc_node_results, urbanev_node_results, metrla_node_results])):
        axes[i][0].set_ylabel('Decoded Value', fontsize=label_size)
        for j, target_type in enumerate(['Location', 'POI', 'Topology', 'Observation']):
            y = (node_results['PCA']['Y'][target_type] - node_results['PCA']['Y'][target_type].mean(axis=0)) / node_results['PCA']['Y'][target_type].std(axis=0)
            y_hat = (node_results['PCA']['Y_hat'][target_type] - node_results['PCA']['Y_hat'][target_type].mean(axis=0)) / node_results['PCA']['Y_hat'][target_type].std(axis=0)
            xmin = ymin = min(y.min(), y_hat.min())
            xmax = ymax = max(y.max(), y_hat.max())
            scores = node_results['PCA']['Score'][target_type]
            axes[i][j].scatter(y.flatten(), y_hat.flatten(), s=1, c='r', label=fr"$R_S={scores['rs']:.2f}, R_P={scores['rp']:.2f},$" + '\n' + fr"$MAE={scores['MAE']:.2f}$")
            axes[i][j].plot([xmin-0.05*(xmax-xmin), xmax+0.05*(xmax-xmin)], [ymin-0.05*(ymax-ymin), ymax+0.05*(ymax-ymin)], lw=1, c='k', ls='--')
            axes[i][j].tick_params(labelsize=label_size-1)
            axes[i][j].set_xlabel('Real Value', fontsize=label_size)
            axes[i][j].set_xlim(xmin-0.05*(xmax-xmin), xmax+0.05*(xmax-xmin))
            axes[i][j].set_ylim(ymin-0.05*(ymax-ymin), ymax+0.05*(ymax-ymin))
            axes[i][j].legend(loc='upper left', fontsize=label_size-1, handlelength=0.5, handletextpad=0.3, fancybox=False, frameon=True)
            axes[i][j].set_title(f'({chr(97+i*4+j)}). {target_type}, {dataset_name}', y=-0.33, fontsize=label_size)
    
    plt.show()
```

```python
from matplotlib.dates import DateFormatter
hh_mm = DateFormatter('%H:%M')

time_index = pd.date_range(start='2000-01-01 00:00', end='2000-01-01 23:55', freq='5min')

dataset_names = ['BART, 1h', 'SUBWAY-MTA, 1h', 'UrbanEV, 1h', 'METR-LA, 5min']
time_step_nums = [24] * 3 + [288]

label_size = 14
with plt.style.context(['science', 'grid', 'no-latex']):
    fig, axes = plt.subplots(2, 4, figsize=(15,6), layout='compressed')
    for row in axes:
        for ax in row:
            ax.tick_params(labelsize=label_size-1)

    ## Time of Day
    i, target_type, target_name = 0, 'Time', 'Time of Day'
    axes[i][0].set_ylabel('Cosine Embedding', fontsize=label_size)
    for j, (dataset_name, time_step_num, time_results) in enumerate(zip(dataset_names, time_step_nums, [bart_time_results, nyc_time_results, urbanev_time_results, metrla_time_results])):
        x_time_index = time_index[::288//time_step_num]
        axes[i][j].plot(x_time_index, time_results['PCA']['Y'][target_type][:time_step_num,1], lw=1, label='Real')
        axes[i][j].plot(x_time_index, time_results['PCA']['Y_hat'][target_type][:time_step_num,1], lw=1, ls='--', label='Linearly Decoded')
        axes[i][j].set_xlim(x_time_index[0], x_time_index[-1])
        axes[i][j].set_xticks(time_index[24::48])
        axes[i][j].xaxis.set_major_formatter(hh_mm)
        axes[i][j].set_title(f'({chr(97+i*4+j)}). {target_name}, {dataset_name}', y=-0.2, fontsize=label_size)

    ## Weekly typical observations
    i, target_type, target_name = 1, 'Observation', 'Weekly Trend'
    ylabels = ['Ridership (pax/h)', 'Ridership (pax/h)', 'Occupancy (%)', 'Speed (km/h)']
    for j, (dataset_name, time_step_num, ylabel, time_results) in enumerate(zip(dataset_names, time_step_nums, ylabels,
                                                                                [bart_time_results, nyc_time_results, urbanev_time_results, metrla_time_results])):
        axes[i][j].plot(time_results['PCA']['Y'][target_type].mean(axis=1), lw=1, label='Real')
        axes[i][j].plot(time_results['PCA']['Y_hat'][target_type].mean(axis=1), lw=1, ls='--', label='Linearly Decoded')
        axes[i][j].set_xlim(0, 7*time_step_num-1)
        axes[i][j].set_xticks(np.arange(0, 7) * time_step_num)
        axes[i][j].set_xticklabels(['Mon.', 'Tue.', 'Wed.', 'Thu.', 'Fri.', 'Sat.', 'Sun.'])
        axes[i][j].set_ylabel(ylabel, fontsize=label_size)
        axes[i][j].set_title(f'({chr(97+i*4+j)}). {target_name}, {dataset_name}', y=-0.2, fontsize=label_size)

    lines, labels = axes[-1][-1].get_legend_handles_labels()
    fig.legend(lines, labels, loc='upper center', ncol=2, fontsize=label_size, fancybox=False, frameon=False, bbox_to_anchor=(0.5, 1.08))

    plt.show()
```
