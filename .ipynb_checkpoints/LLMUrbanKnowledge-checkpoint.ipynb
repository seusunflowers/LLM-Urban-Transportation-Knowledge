This script facilitates the extraction of urban-related knowledge and hidden-state embeddings from LLMs (using Qwen-3-14B as an example) across multiple transit and traffic datasets. It generates textual descriptions and captures numerical representations (embeddings) for spatial, temporal, and daily urban patterns. The LLM inference relies on class `HuggingFaceLlmTools`, which encapsulates following **key functions**:
- **Model-Specific Tokenization** (`tokenize_llm_inputs`): Formats input text into model-specific prompt templates, such as Llama 3.1 headers or Qwen 3 chat templates with thinking mode enabled.
- **Response Decoding** (`decode_llm_outputs`): Post-processes raw token sequences. For Qwen 3, it specifically extracts and separates the reasoning (thinking_content) from the final content using the </think> token.
- **Embedding Extraction** (`get_llm_outputs`): Generates the model response and extracts the hidden-state embeddings for the last generated token.
- **Automated Batch Processing** (`enumerate_llm_outputs`): Iteratively processes spatial, temporal, and daily queries for entire datasets.

```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame

import torch
from tools.hugging_face_tools import HuggingFaceLlmTools
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
BATCH_SIZE = 64

MODEL_DIR = {
    'qwen3-14b': '/path/to/the/model',
}
LLM_GENERATION_PARAMS = {
    'qwen3-14b-thinking': {'temperature':0.6, 'top_p':0.95, 'top_k':20, 'min_p':0, 'max_new_tokens':32768},
    'qwen3-14b-nothinking': {'temperature':0.7, 'top_p':0.8, 'top_k':20, 'min_p':0, 'max_new_tokens':32768},
}
THINKING = True
```

```python
def get_llm_node_time_day_embeddings(llm_tools:HuggingFaceLlmTools, spatial_queries:list, temporal_queries:list, daily_queries:list, spatial_items:list, temporal_items:list, daily_items:list):
    """
    Generates LLM-based textual descriptions and hidden-state embeddings for spatial, temporal, and day-of-week queries.

    Args:
        llm_tools: Initialized HuggingFaceLlmTools instance used for inference.
        spatial_queries: Queries describing spatial units (e.g., nodes or regions).
        temporal_queries: Queries describing temporal intervals.
        daily_queries: Queries describing days of the week.
        spatial_items: Labels corresponding to the spatial queries.
        temporal_items: Labels corresponding to the temporal queries.
        daily_items: Labels corresponding to the daily queries.

    Returns:
        A tuple containing:
            - node_description_llm (DataFrame): Generated descriptions for spatial queries.
            - node_hidden_states_llm (ndarray): Hidden-state embeddings for spatial queries.
            - time_description_llm (DataFrame): Generated descriptions for temporal queries.
            - time_hidden_states_llm (ndarray): Hidden-state embeddings for temporal queries.
            - day_description_llm (DataFrame): Generated descriptions for daily queries.
            - day_hidden_states_llm (ndarray): Hidden-state embeddings for day-of-week queries.
    """
    ## Spatial Queries
    node_description_llm, node_hidden_states_llm = llm_tools.enumerate_llm_outputs(queries=spatial_queries, no_grad=True, enable_thinking=THINKING, output_hidden_states=True,
                                                                                   **LLM_GENERATION_PARAMS[f'{LLM_NAME}-{"" if THINKING else "no"}thinking'])
    ## Temporal Queries
    time_description_llm, time_hidden_states_llm = llm_tools.enumerate_llm_outputs(queries=temporal_queries, no_grad=True, enable_thinking=THINKING, output_hidden_states=True,
                                                                                   **LLM_GENERATION_PARAMS[f'{LLM_NAME}-{"" if THINKING else "no"}thinking'])
    ## Daily Queries
    day_description_llm, day_hidden_states_llm = llm_tools.enumerate_llm_outputs(queries=daily_queries, no_grad=True, enable_thinking=THINKING, output_hidden_states=True,
                                                                                 **LLM_GENERATION_PARAMS[f'{LLM_NAME}-{"" if THINKING else "no"}thinking'])
    for description, items, item_type in zip([node_description_llm, time_description_llm, day_description_llm], [spatial_items, temporal_items, daily_items], ['Node', 'Time', 'Day']):
        description[item_type] = items
    return node_description_llm, node_hidden_states_llm, time_description_llm, time_hidden_states_llm, day_description_llm, day_hidden_states_llm
```

# BART
The BART dataset includes the hourly ridership of **50 stations** in the **San Francisco Bay Area’s** rapid transit system in **2024**. Each station serves as a basic spatial unit. Since metro stations are commonly the local landmarks, their names are provided to pretrained LLM agents to obtain spatial heterogeneity insights.

```python
RAW_DATASET_DIR = 'datasets/bart'

LLM_NAME = 'qwen3-14b'
LLM_NAME_ = '_'.join(LLM_NAME.split('-'))
llm_tools = HuggingFaceLlmTools(llm_name=LLM_NAME, model_dir=MODEL_DIR[LLM_NAME], tokenizer_dir=MODEL_DIR[LLM_NAME], tokenizer_params={'model_max_length':8192}, device='cuda')

## prepare queries
raw_data = np.load(f'{RAW_DATASET_DIR}/station-based-st-graph.npz', allow_pickle=True)
station_ids = raw_data['station_ids']
station_info = pd.read_excel(f'{RAW_DATASET_DIR}/station-names.xls', index_col=0)
station_info = station_info.loc[station_ids]
spatial_queries = (f'Please provide details about the location, train operations, ridership demand, and landmarks nearby about the {station} Station of the San Francisco Bay Area Rapid Transit System' 
                   for station in station_info['Station Name'])                                                                                                                                       # Spatial Queries
temporal_queries = (f'San Francisco Bay Area Rapid Transit System during {start_time:02d}:00-{end_time:02d}:00, service and ridership ' for start_time, end_time in zip(range(24), range(1, 25)))     # Temporal Queries
daily_queries = (f'San Francisco Bay Area Rapid Transit System on a typical {day}, service and ridership ' for day in ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']) # Daily Queries
## obtain llm knowledge
node_description_llm, node_hidden_states_llm, time_description_llm, time_hidden_states_llm, day_description_llm, day_hidden_states_llm = get_llm_node_time_day_embeddings(
    llm_tools, spatial_queries, temporal_queries, daily_queries, spatial_items=station_info.index.copy(), temporal_items=np.arange(24),
    daily_items=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
## save llm knowledge
for description, description_type in zip([node_description_llm, time_description_llm, day_description_llm], ['node', 'time', 'day']):
    description.to_csv(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/{description_type}_description_{LLM_NAME_}.csv.zip', compression='zip', encoding='utf-8-sig')
np.savez_compressed(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/hidden_states_{LLM_NAME_}.npz', node=node_hidden_states_llm, time=time_hidden_states_llm, day=day_hidden_states_llm)
```

# Demand-QD
The Demand-QD dataset includes **trip demand data collected from May 2 to May 14, 2021, across the central urban area of Qingdao, China**. A total of **84 subdistricts** serve as the basic spatial units for trip aggregation, and their names are provided to the LLM to obtain spatial heterogeneity features. The trip data have a temporal resolution of 5 minutes, whereas the LLM is queried to infer temporal features at 15-minute intervals. **The prompts are written in Chinese** because the dataset describes a transportation system in China. Notably, **the trip data are private and were collected from Baidu Map, a mainstream location-based service provider in China. The collected data cover only a subset of the overall travel demand and are inaccessible to pretrained LLMs.**

```python
RAW_DATASET_DIR = 'datasets/qingdao'

LLM_NAME = 'qwen3-14b'
LLM_NAME_ = '_'.join(LLM_NAME.split('-'))
llm_tools = HuggingFaceLlmTools(llm_name=LLM_NAME, model_dir=MODEL_DIR[LLM_NAME], tokenizer_dir=MODEL_DIR[LLM_NAME], tokenizer_params={'model_max_length':8192}, device='cuda')

## prepare queries
raw_data = np.load(f'{RAW_DATASET_DIR}/districttrip-st-graph.npz', allow_pickle=True)
spatial_queries = (f'中国山东省青岛市{district}的经济水平、人口规模、土地利用、产业结构与交通出行需求强度。' for district in raw_data['districtname'])                                         # Spatial Queries
temporal_queries = (f'请简要介绍中国山东省青岛市中心城区{time}的交通出行需求。' for time in pd.date_range(start='2000-01-01 00:00', end='2000-01-01 23:55', freq='15min').strftime('%H:%M')) # Temporal Queries
daily_queries = (f'请简要介绍中国山东省青岛市中心城区一个典型{day}的交通出行需求。' for day in ['周一', '周二', '周三', '周四', '周五', '周六', '周日', '节假日'])                               # Daily Queries
## obtain llm knowledge
node_description_llm, node_hidden_states_llm, time_description_llm, time_hidden_states_llm, day_description_llm, day_hidden_states_llm = get_llm_node_time_day_embeddings(
    llm_tools, spatial_queries, temporal_queries, daily_queries, spatial_items=raw_data['districtname'].copy(),
    temporal_items=pd.date_range(start='2000-01-01 00:00', end='2000-01-01 23:55', freq='15min').strftime('%H:%M'), daily_items=['周一', '周二', '周三', '周四', '周五', '周六', '周日', '节假日'])
## save llm knowledge
for description, description_type in zip([node_description_llm, time_description_llm, day_description_llm], ['node', 'time', 'day']):
    description.to_csv(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/{description_type}_description_{LLM_NAME_}.csv.zip', compression='zip', encoding='utf-8-sig')
np.savez_compressed(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/hidden_states_{LLM_NAME_}.npz', node=node_hidden_states_llm, time=time_hidden_states_llm, day=day_hidden_states_llm)
```

# UrbanEV
The UrbanEV dataset includes hourly electric vehicle charging occupancies from 17,532 piles **between September 1, 2022 and February 28, 2023**, in **Shenzhen City**, Guangdong Province, China. A total of **275 traffic analysis zones (TAZs)** serve as basic spatial units for occupancy aggregation. Since the TAZs are randomly labeled, the authors provide all the names of bus and metro stations within the corresponding TAZ to LLM agents for spatial heterogeneity insights. **The raw queries are prepared in Chinese for the dataset describing a transportation system in China.**

```python
RAW_DATASET_DIR = 'datasets/UrbanEV'

LLM_NAME = 'qwen3-14b'
LLM_NAME_ = '_'.join(LLM_NAME.split('-'))
llm_tools = HuggingFaceLlmTools(llm_name=LLM_NAME, model_dir=MODEL_DIR[LLM_NAME], tokenizer_dir=MODEL_DIR[LLM_NAME], tokenizer_params={'model_max_length':8192}, device='cuda')

## prepare queries
pd.read_csv(f'{RAW_DATASET_DIR}/UrbanEVDataset/20220901-20230228_zone-cleaned-aggregated/zone-stations.csv', usecols=[2,12,14])
spatial_queries = (f'中国广东省深圳市{region}附近区域的土地利用、居民组成、交通出行模式、新能源汽车友好度、新能源汽车充电需求与服务水平。' 
                   for region in (sub_points['DISTRICT'].iloc[0] + '、'.join(sub_points['name']) for _, sub_points in zones_points.groupby('TAZID')))                # Spatial Queries
temporal_queries = (f'中国广东省深圳市{start_time:02d}:00-{end_time:02d}:00间的新能源汽车充电需求与服务水平。' for start_time, end_time in zip(range(24), range(1, 25))) # Temporal Queries
daily_queries = (f'中国广东省深圳市一个典型{day}的新能源汽车充电需求与服务水平。' for day in ['周一', '周二', '周三', '周四', '周五', '周六', '周日', '节假日'])             # Daily Queries
## obtain llm knowledge
node_description_llm, node_hidden_states_llm, time_description_llm, time_hidden_states_llm, day_description_llm, day_hidden_states_llm = get_llm_node_time_day_embeddings(
    llm_tools, spatial_queries, temporal_queries, daily_queries, spatial_items=np.sort(zones_points['TAZID'].unique()), temporal_items=np.arange(24),
    daily_items=['周一', '周二', '周三', '周四', '周五', '周六', '周日', '节假日'])
## save llm knowledge
for description, description_type in zip([node_description_llm, time_description_llm, day_description_llm], ['node', 'time', 'day']):
    description.to_csv(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/{description_type}_description_{LLM_NAME_}.csv.zip', compression='zip', encoding='utf-8-sig')
np.savez_compressed(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/hidden_states_{LLM_NAME_}.npz', node=node_hidden_states_llm, time=time_hidden_states_llm, day=day_hidden_states_llm)
```

# METR-LA
The METR-LA dataset contains 5-minute average speed data collected from **207 urban highway sensors from March to June in 2012**, released by the **Los Angeles** Metropolitan Transportation Authority. Each sensor serves as a basic spatial unit, labeled with the name of nearby landmark. Therefore, the sensors’ names with their located highways are provided to pretrained LLM agents to obtain spatial heterogeneity insights. The temporal details are inquired at a 15-minute interval.

```python
RAW_DATASET_DIR = 'datasets/metr-la'

LLM_NAME = 'qwen3-14b'
LLM_NAME_ = '_'.join(LLM_NAME.split('-'))
llm_tools = HuggingFaceLlmTools(llm_name=LLM_NAME, model_dir=MODEL_DIR[LLM_NAME], tokenizer_dir=MODEL_DIR[LLM_NAME], tokenizer_params={'model_max_length':8192}, device='cuda')

## prepare queries
station_info = pd.read_csv(f'{RAW_DATASET_DIR}/graph_sensor_metadata.csv', usecols=[1, 5, 12, 13, 14, 15])
fwy_names = ', '.join(set(station_info['FwyName']))
city_names = 'Burbank, Glendale, La Canada-Flintridge, and Los Angeles'
spatial_queries = (f'Please provide details about the location, traffic demand pattern, and traffic condition on a segment of the {fwy_name + '-' + fwy_dir} Freeway around {det} in {city}, {county} County, California, USA.' 
                   for _, _, fwy_dir, det, county, fwy_name, city in station_info.itertuples())                                                # Spatial Queries
temporal_queries = (f'Typical traffic demands and conditions at {time} on {fwy_names} in {city_names}, Los Angeles County, California, USA.' 
                    for time in pd.date_range(start='2000-01-01 00:00', end='2000-01-01 23:55', freq='15min').strftime('%H:%M'))               # Temporal Queries
daily_queries = (f'Traffic demands and conditions on a typical {day} on {fwy_names} in {city_names}, Los Angeles County, California, USA.' 
                 for day in ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])                                    # Daily Queries
## obtain llm knowledge
node_description_llm, node_hidden_states_llm, time_description_llm, time_hidden_states_llm, day_description_llm, day_hidden_states_llm = get_llm_node_time_day_embeddings(
    llm_tools, spatial_queries, temporal_queries, daily_queries, spatial_items=station_info['sensor_id'].copy(), 
    temporal_items=pd.date_range(start='2000-01-01 00:00', end='2000-01-01 23:55', freq='15min').strftime('%H:%M'), daily_items=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
## save llm knowledge
for description, description_type in zip([node_description_llm, time_description_llm, day_description_llm], ['node', 'time', 'day']):
    description.to_csv(f'{RAW_DATASET_DIR}/llm-description/description-{LLM_NAME}/{description_type}_description_{LLM_NAME_}.csv.zip', compression='zip', encoding='utf-8-sig')
np.savez_compressed(f'{RAW_DATASET_DIR}/llm-description/description-{LLM_NAME}/hidden_states_{LLM_NAME_}.npz', node=node_hidden_states_llm, time=time_hidden_states_llm, day=day_hidden_states_llm)
```

# subway-mta
The SUBWAY-MTA dataset provides hourly ridership data for **428 subway stations** in the **New York City** metropolitan area in **2024**. Details for obtaining LLMs’ knowledge are consistent with those for the BART dataset.

```python
RAW_DATASET_DIR = 'datasets/subway-mta'

LLM_NAME = 'qwen3-14b'
LLM_NAME_ = '_'.join(LLM_NAME.split('-'))
llm_tools = HuggingFaceLlmTools(llm_name=LLM_NAME, model_dir=MODEL_DIR[LLM_NAME], tokenizer_dir=MODEL_DIR[LLM_NAME], tokenizer_params={'model_max_length':8192}, device='cuda')

## prepare queries
station_info = pd.read_csv(f'{RAW_DATASET_DIR}/MTA_Subway_Hourly_Ridership__2020-2024_20250317.zip', usecols=[3, 4])
station_info.drop_duplicates(inplace=True)
station_info.set_index('station_complex', inplace=True)
spatial_queries = (f'Please provide details about the location, train operations, ridership demand, and landmarks nearby about the {station} Station in {borough} of the New York City Subway System.' 
                   for station, borough in zip(station_info.index, station_info['borough']))                                                                                          # Spatial Queries
temporal_queries = (f'New York City Subway System during {start_time:02d}:00-{end_time:02d}:00, service and ridership.' for start_time, end_time in zip(range(24), range(1, 25)))     # Temporal Queries
daily_queries = (f'New York City Subway System on a typical {day}, service and ridership.' for day in ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']) # Daily Queries
## obtain llm knowledge
node_description_llm, node_hidden_states_llm, time_description_llm, time_hidden_states_llm, day_description_llm, day_hidden_states_llm = get_llm_node_time_day_embeddings(
    llm_tools, spatial_queries, temporal_queries, daily_queries, spatial_items=station_info.index.copy(), temporal_items=np.arange(24),
    daily_items=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
## save llm knowledge
for description, description_type in zip([node_description_llm, time_description_llm, day_description_llm], ['node', 'time', 'day']):
    description.to_csv(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/{description_type}_description_{LLM_NAME_}.csv.zip', compression='zip', encoding='utf-8-sig')
np.savez_compressed(f'{RAW_DATASET_DIR}/description-{LLM_NAME}/hidden_states_{LLM_NAME_}.npz', node=node_hidden_states_llm, time=time_hidden_states_llm, day=day_hidden_states_llm)
```
